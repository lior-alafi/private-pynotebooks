```python
movies = pd.read_csv('../DATA/ml-20m/movies.csv')
movies['genres'] = movies['genres'].str.split('|')
movies_dict = {}
generes_dict = {}


genere_id = 0
genere_max_len = 0
for r in movies.values:
    movie_name = r[1].lower()
    if movie_name not in movies_dict:
        movies_dict[movie_name]= r[0]
    genere_max_len = np.max([genere_max_len,len(r[2])])
  
    
    for gen in r[2]:
        gen= gen.lower()
        if gen not in generes_dict:
            generes_dict[gen] = genere_id
            genere_id += 1



print(movies_dict)
print(generes_dict)
print(genere_max_len)
```
