# Model Loading (assuming ex3 ran and model was saved)

in order to load the model we first need to recreate the model itself:


```python
import torch
import torch.nn as nn
import numpy as np
```

```python
model = nn.Sequential(
    nn.Linear(30,1),
    nn.Sigmoid()
)
```

## load the model

```python
model.load_state_dict(torch.load('../MODELS/cancer_classification.pt'))
```

now let's bring the dataset again and eval

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer

dataset = load_breast_cancer()
```

```python
dataset.keys()
```

```python
X,y = dataset.data,dataset.target
y = y.reshape(-1,1)
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25)
X_train.shape
```

```python
dataset.feature_names
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
X_train = torch.from_numpy(X_train.astype(np.float32))
X_test = torch.from_numpy(X_test.astype(np.float32))
y_train = torch.from_numpy(y_train.astype(np.float32))
y_test = torch.from_numpy(y_test.astype(np.float32))
```

```python
with torch.no_grad():
    test_pred = model(X_test)
    train_pred = model(X_train)

    test_acc = np.mean(np.round(test_pred.numpy()) == y_test.numpy())
    train_acc = np.mean(np.round(train_pred.numpy()) == y_train.numpy())

print(f'Train Accuracy: {train_acc} Test accuracy: {test_acc}')
```
