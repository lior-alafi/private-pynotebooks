```python
import numpy as np

# Define dimensions
N, T, D = 6, 5, 3  # Batch size, sequence length, input size
L, M = 2, 4        # Number of layers, hidden size

# Input at a single time step (X_t is taken per time step)
X_t = np.random.randn(N, D)  # Shape: (N, D)

# Weight matrices
W_xh = np.random.randn(D, M)  # Input-to-hidden weights (D, M)
W_hh = np.random.randn(M, M)  # Hidden-to-hidden weights (M, M)

# Previous hidden state (for all layers)
h_t_minus_1 = np.random.randn(L, N, M)  # Shape: (L, N, M)

# Step 1: Compute input-to-hidden transformation
# Matrix multiplication X_t @ W_xh for each batch element
X_t_W_xh = X_t @ W_xh  # Shape: (N, M)

# Step 2: Expand (N, M) to (L, N, M) for layer-wise addition
# We use np.newaxis to add the new dimension and then broadcast
X_t_W_xh_expanded = X_t_W_xh[np.newaxis, :, :]  # Shape: (1, N, M)
X_t_W_xh_expanded = np.tile(X_t_W_xh_expanded, (L, 1, 1))  # Shape: (L, N, M)

# Step 3: Compute hidden-to-hidden transformation
# Batch-wise matrix multiplication h_t_minus_1 @ W_hh for each layer and batch element
h_t_minus_1_W_hh = np.einsum("lnm,mk->lnk", h_t_minus_1, W_hh)  # Shape: (L, N, M)

# Step 4: Add both terms
pre_activation = X_t_W_xh_expanded + h_t_minus_1_W_hh  # Shape: (L, N, M)

# Output the shape to confirm correctness
print("Final pre-activation shape:", pre_activation.shape)  # Expected: (L, N, M)
```
